```python
import os
import html2text
from boilerpy3.extractors import ArticleExtractor
import justext
from bs4 import BeautifulSoup
import markdownify
```

```python
extractor = ArticleExtractor()
def extract_main(content, as_markdown=True):
    """
    content: URL, HTML, or text your ArticleExtractor can handle.
    as_markdown: True -> return Markdown; False -> return sanitized HTML.
    """
    try:
        html = extractor.get_marked_html(content)  # main article HTML
    except Exception:
        html = content  # fallbackL
    return html
```

```python
def convert_html_folder_to_markdown(input_folder, output_folder, ignore_images=True, ignore_links=True):
    """
    Convert all .html files in input_folder (recursively) to markdown files saved in output_folder.

    Args:
        input_folder (str): Path to folder containing HTML files.
        output_folder (str): Path to folder where markdown files will be saved.
        ignore_images (bool): Whether to skip images in conversion. Default True.
        ignore_links (bool): Whether to skip links in conversion. Default True.
    """
    os.makedirs(output_folder, exist_ok=True)

    converter = html2text.HTML2Text()
    converter.ignore_images = ignore_images
    converter.ignore_links = ignore_links

    for root_dir, _, files in os.walk(input_folder):
        for file in files:
            if file.lower().endswith(".html"):
                html_path = os.path.join(root_dir, file)
                with open(html_path, "r", encoding="utf-8") as f:
                    html_content = f.read()

                # Get sanitized HTML, then use html2text to honor ignore_* flags
                cleaned_html= extract_main(html_content)
                markdown = converter.handle(cleaned_html)
                #print(markdown)
                #print(cleaned_html)

                base_name = os.path.splitext(file)[0]
                md_path = os.path.join(output_folder, base_name + ".md")
                with open(md_path, "w", encoding="utf-8") as f:
                    f.write(markdown)

    print("Converted all html files")
```

```python
#queries
input_folder = "/content/drive/MyDrive/MIT807/data/ConsumerQueries/html/"
output_folder = "/content/drive/MyDrive/MIT807/data/raw/consumer_queries/"
converter = html2text.HTML2Text()
converter.ignore_links = True  # Keep links if you want
converter.ignore_images = True  # Skip images
convert_html_folder_to_markdown(input_folder, output_folder)
```
