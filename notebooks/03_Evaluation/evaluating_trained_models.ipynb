# Evaluate All Training Models

## BioMistral

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
results_path = "/content/drive/MyDrive/MIT807/biomistral_qlora/trainer_state.json"
qlora = pd.read_json(results_path).sort_index()

# Create two subplots next to each other
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# -------------------------
# LEFT: Accuracy
# -------------------------
sns.lineplot(data=qlora, x="step", y="mean_token_accuracy",
             ax=axes[0], label="Training Accuracy", color="tab:blue")
sns.lineplot(data=qlora, x="step", y="eval_mean_token_accuracy",
             ax=axes[0], label="Validation Accuracy", color="tab:green")

# Highlight the maximum validation accuracy
if "eval_mean_token_accuracy" in qlora.columns and qlora["eval_mean_token_accuracy"].notnull().any():
    max_idx = qlora["eval_mean_token_accuracy"].idxmax()
    max_point = qlora.loc[max_idx, ["step", "eval_mean_token_accuracy"]]
    axes[0].scatter(max_point["step"], max_point["eval_mean_token_accuracy"],
                    color="black", s=40, zorder=5)
    axes[0].text(max_point["step"], max_point["eval_mean_token_accuracy"],
                 f"{max_point['eval_mean_token_accuracy']:.3f}",
                 color="black", fontsize=8, va="bottom")

axes[0].set_title("Accuracy")
axes[0].set_xlabel("Training Step")
axes[0].set_ylabel("Mean Token Accuracy")
axes[0].legend()
axes[0].grid(True, linestyle="--", alpha=0.6)

# -------------------------
# RIGHT: Loss
# -------------------------
sns.lineplot(data=qlora, x="step", y="loss",
             ax=axes[1], label="Training Loss", color="tab:red")
sns.lineplot(data=qlora, x="step", y="eval_loss",
             ax=axes[1], label="Validation Loss", color="tab:orange")

# Highlight minimum validation loss
if "eval_loss" in qlora.columns and qlora["eval_loss"].notnull().any():
    min_idx = qlora["eval_loss"].idxmin()
    min_point = qlora.loc[min_idx, ["step", "eval_loss"]]
    axes[1].scatter(min_point["step"], min_point["eval_loss"],
                    color="black", s=40, zorder=5)
    axes[1].text(min_point["step"], min_point["eval_loss"],
                 f"{min_point['eval_loss']:.3f}",
                 color="black", fontsize=8, va="bottom")

axes[1].set_title("Loss")
axes[1].set_xlabel("Training Step")
axes[1].set_ylabel("Loss")
axes[1].legend()
axes[1].grid(True, linestyle="--", alpha=0.6)

# -------------------------
# Final layout tweaks
# -------------------------
plt.suptitle("BioMistral QLoRA Training & Validation Loss & Accuracy", fontsize=14)
plt.tight_layout()
plt.show()
```

## QLoRA

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
results_path = "/content/drive/MyDrive/MIT807/biomistral_prompt_tuning/trainer_state.json"
qlora_v1 = pd.read_json(results_path).sort_index()
qlora_v1['Model'] = 'BioMistral-7B'
# results_path = "/content/drive/MyDrive/MIT807/biomedlm_qlora/trainer_state.json"
# qlora_v2 = pd.read_json(results_path).sort_index()
# qlora_v2['Model'] = 'BioMedLM-2.7B'
results_path = "/content/drive/MyDrive/MIT807/qwen_prompt_tuning/trainer_state.json"
qlora_v3 = pd.read_json(results_path).sort_index()
qlora_v3['Model'] = 'Qwen3-8B'
```

```python
qlora = pd.concat([qlora_v1,qlora_v3])
qlora.head()
```

```python
qlora = qlora.reset_index(drop=True)

# Create two subplots next to each other
fig, axes = plt.subplots(1, 2, figsize=(9, 7))

# Prepare data for Accuracy plot
df_accuracy = qlora.melt(
    id_vars=['step', 'Model'],
    value_vars=['mean_token_accuracy', 'eval_mean_token_accuracy'],
    var_name='Metric Type',
    value_name='Accuracy'
)
# Rename metric types for better legend
df_accuracy['Metric Type'] = df_accuracy['Metric Type'].replace({
    'mean_token_accuracy': 'Training Accuracy',
    'eval_mean_token_accuracy': 'Validation Accuracy'
})

# -------------------------
# LEFT: Accuracy
# -------------------------
sns.lineplot(data=df_accuracy, x="step", y="Accuracy",
             hue="Model", style="Metric Type", ax=axes[0])

axes[0].set_title("Accuracy")
axes[0].set_xlabel("Training Step")
axes[0].set_ylabel("Mean Token Accuracy")
axes[0].grid(True, linestyle="--", alpha=0.6)

# Prepare data for Loss plot
df_loss = qlora.melt(
    id_vars=['step', 'Model'],
    value_vars=['loss', 'eval_loss'],
    var_name='Metric Type',
    value_name='Loss'
)
# Rename metric types for better legend
df_loss['Metric Type'] = df_loss['Metric Type'].replace({
    'loss': 'Training Loss',
    'eval_loss': 'Validation Loss'
})

# -------------------------
# RIGHT: Loss
# -------------------------
sns.lineplot(data=df_loss, x="step", y="Loss",
             hue="Model", style="Metric Type", ax=axes[1])

axes[1].set_title("Loss")
axes[1].set_xlabel("Training Step")
axes[1].set_ylabel("Loss")
axes[1].grid(True, linestyle="--", alpha=0.6)

# -------------------------
# Final layout tweaks
# -------------------------
plt.suptitle("Prompt Tuning Training & Validation Loss & Accuracy", fontsize=14)
plt.tight_layout()
plt.show()
```
